```python
import pandas as pd

# Week 7 - Outlier Detection using IQR

INPUT_FILE = "week6_sold_feature_engineered.csv"

OUTPUT_FLAGGED = "sold_week7_flagged.csv"
OUTPUT_FILTERED = "sold_week7_filtered.csv"

df = pd.read_csv(INPUT_FILE)
# Columns
numeric_cols = [
    "ClosePrice",
    "LivingArea",
    "DaysOnMarket"
]
```

```python
# Create IQR
for col in numeric_cols:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    flag_col = f"{col}_Outlier"

    df[flag_col] = (
        (df[col] < lower) |
        (df[col] > upper)
    )

    print(f"\n{col}")
    print(f"Q1: {Q1:,.2f}")
    print(f"Q3: {Q3:,.2f}")
    print(f"IQR: {IQR:,.2f}")
    print(f"Lower Bound: {lower:,.2f}")
    print(f"Upper Bound: {upper:,.2f}")
    print(f"Outliers: {df[flag_col].sum()}")


# Save
df.to_csv(OUTPUT_FLAGGED, index=False)
```

```python
# Create filtered analysis dataset
# Remove rows flagged
flag_columns = [f"{c}_Outlier" for c in numeric_cols]

filtered_df = df.loc[
    ~df[flag_columns].any(axis=1)
].copy()

filtered_df.to_csv(OUTPUT_FILTERED, index=False)

```

```python
# Dataset comparison

print("Dataset Size Comparison")

print(f"Original rows : {len(df):,}")
print(f"Filtered rows : {len(filtered_df):,}")
print(f"Rows removed  : {len(df)-len(filtered_df):,}")

print("Median Comparison")

comparison = pd.DataFrame({
    "Before": df[numeric_cols].median(),
    "After": filtered_df[numeric_cols].median()
})

print(comparison)
```

```python

# save comparison table
comparison.to_csv("week7_median_comparison.csv")

print("\nFiles saved:")
print(f"  {OUTPUT_FLAGGED}")
print(f"  {OUTPUT_FILTERED}")
print("  week7_median_comparison.csv")
```
